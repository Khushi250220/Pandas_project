```python
# import python libraries

import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt # visualizing data
%matplotlib inline
import seaborn as sns
```

```python
# import csv file
df = pd.read_csv('Diwali Sales Data.csv', encoding= 'unicode_escape')
```

# Exploratory Data Analysis

```python
sns.countplot(x= 'Gender',data =df,hue="Gender",palette='Set2')
```

```python
sns.countplot(x='Age Group',data =df,hue= 'Gender',palette='Set2')
```

```python
sns.countplot(x='State',data =df,hue='Orders',palette='Set2')
sns.set(rc={'figure.figsize':(20,5)})
```

```python
sns.countplot(x='Zone',palette='Set1',hue='Zone',data =df)
sns.set(rc={'figure.figsize':(20,5)})
```

```python
sns.countplot(palette='Set2',x='Marital_Status',data =df,hue='Marital_Status')
sns.set(rc={'figure.figsize':(18,5)})
```

## Occupation

```python
ax = sns.countplot(x='Occupation',data =df,palette='Set2',hue='Occupation')
sns.set(rc={'figure.figsize':(20,5)})
for bar in ax.containers:
    ax.bar_label(bar)
plt.show()
```

## product Category

```python
ax = sns.countplot(x='Product_Category',data =df,palette='Set1',hue='Product_Category')
sns.set(rc={'figure.figsize':(35,5)})
for bar in ax.containers:
    ax.bar_label(bar)
plt.show()
```
